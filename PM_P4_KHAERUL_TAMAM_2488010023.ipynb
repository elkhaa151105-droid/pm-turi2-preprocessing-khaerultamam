```python

import pandas as pd
import numpy as np

data = {
    'usia': [25, 32, np.nan, 45, 28, 51, 38, np.nan, 29, 41],

    'pendapatan': [
        4_000_000, 7_500_000, 5_200_000, 12_000_000, 4_800_000,
        15_000_000, 8_100_000, 6_300_000, np.nan, 9_900_000
    ],

    'pendidikan': [
        'SMA', 'S1', 'SMA', 'S2', 'SMA',
        'S2', 'S1', 'S1', 'SMA', 'S2'
    ],

    'kota': [
        'Bandung', 'Jakarta', 'Bandung', 'Surabaya', 'Jakarta',
        'Surabaya', 'Jakarta', 'Bandung', 'Bandung', 'Jakarta'
    ],

    'membeli': [
        'Tidak', 'Ya', 'Tidak', 'Ya', 'Tidak',
        'Ya', 'Ya', 'Tidak', 'Tidak', 'Ya'
    ]
}

df = pd.DataFrame(data)

# Menampilkan dataset awal
df
```

```python
df['usia'] = df['usia'].fillna(df['usia'].median())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].median())
df.isnull().sum()
```

```python
X = df.drop(columns=['membeli'])
y = df['membeli']

print("X:")
display(X)

print("y:")
display(y)
```

```python
X['pendidikan'] = X['pendidikan'].map({'SMA': 0, 'S1': 1, 'S2': 2})  # ordinal

X = pd.get_dummies(X, columns=['kota'], dtype=int)  # nominal

y = y.map({'Tidak': 0, 'Ya': 1})

print("X setelah encoding:")
display(X)

print("y setelah encoding:")
display(y)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler
num = ['usia', 'pendapatan', 'pendidikan']
sc = StandardScaler()
X_train[num] = sc.fit_transform(X_train[num])
X_test[num] = sc.transform(X_test[num])
```

```python
#mengecek nilai hilang
print("Nilai hilang pada X_train:")
print(X_train.isnull().sum())

print("\nNilai hilang pada X_test:")
print(X_test.isnull().sum())
```

```python
#mengecek tipe data
print("Tipe data X_train:")
print(X_train.dtypes)

print("\nTipe data X_test:")
print(X_test.dtypes)
```

```python
#melihat hasil scaling
print("Statistik X_train setelah scaling:")
display(X_train.describe())
```

```python
print("X_train:")
display(X_train)

print("X_test:")
display(X_test)

print("y_train:")
display(y_train)

print("y_test:")
display(y_test)
```

Mengapa fit_transform pada latih dan transform pada uji?

`fit_transform()` digunakan pada data latih karena scaler perlu mempelajari parameter dari data latih terlebih dahulu. Pada StandardScaler, parameter tersebut berupa nilai mean dan standar deviasi.

Setelah parameter diperoleh dari data latih, data uji hanya diproses menggunakan `transform()` dengan parameter yang sama. Data uji tidak boleh digunakan untuk melakukan `fit` karena hal tersebut dapat menyebabkan data leakage, yaitu informasi dari data uji masuk ke dalam proses pembelajaran preprocessing.

Oleh karena itu, proses yang benar adalah:

X_train → fit_transform()

X_test → transform()

Dengan cara ini, preprocessing pada data uji tetap menggunakan informasi yang diperoleh dari data latih saja.

Refleksi
1. Mengapa urutan “split dulu, baru scaling” penting?

  Urutan “split dulu, baru scaling” penting untuk mencegah terjadinya data leakage. Data harus dibagi terlebih dahulu menjadi data latih dan data uji. Setelah itu, scaler di-fit hanya menggunakan data latih.
  Pada data latih digunakan fit_transform() karena scaler perlu mempelajari parameter dari data latih. Sedangkan pada data uji hanya digunakan transform() menggunakan parameter yang telah diperoleh dari data latih.
  Dengan urutan tersebut, informasi dari data uji tidak digunakan dalam proses preprocessing data latih.

2. Kapan label encoding dan kapan one-hot encoding?

Label encoding digunakan ketika kategori memiliki urutan atau tingkatan tertentu. Pada praktikum ini, kolom `pendidikan` menggunakan ordinal encoding karena memiliki tingkatan SMA, S1, dan S2, yang kemudian dipetakan menjadi 0, 1, dan 2.
One-hot encoding digunakan untuk kategori nominal yang tidak memiliki urutan. Pada praktikum ini, kolom `kota` menggunakan one-hot encoding karena kategori Bandung, Jakarta, dan Surabaya tidak memiliki tingkatan tertentu.


3. Apa perbedaan normalisasi dan standarisasi?

Normalisasi merupakan proses penskalaan data ke rentang tertentu, misalnya 0 sampai 1. Pada latihan mandiri, proses ini dilakukan menggunakan `MinMaxScaler`.
Standardisasi merupakan proses mengubah data berdasarkan rata-rata dan standar deviasi sehingga data memiliki skala yang terstandarisasi. Pada praktikum utama, proses ini dilakukan menggunakan `StandardScaler`.
Dengan demikian, normalisasi dan standardisasi sama-sama digunakan untuk melakukan penskalaan fitur, tetapi metode dan hasil transformasinya berbeda.
